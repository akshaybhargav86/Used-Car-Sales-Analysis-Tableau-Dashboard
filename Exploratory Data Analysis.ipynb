Which Company is selling expensive cars?

```python
top_make = df['make'].value_counts().head(10).index

sns.barplot(x='make', y='sellingprice', data=df[df['make'].isin(top_make)])
plt.title('Top Make vs Selling Price')
plt.xticks(rotation=45)
plt.show()
```

Most Popular cars?

```python
top_model = df['model'].value_counts().head(10).index

sns.countplot(x='model', data=df[df['model'].isin(top_model)])
plt.title('Top Selling Car Models')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.scatterplot(x='odometer', y='sellingprice', data=df, alpha=0.5)
plt.title('Odometer vs Selling Price')
plt.show()
```

```python
trans_data = df[df['transmission'].isin(['automatic','manual'])]
sns.countplot(x='transmission',data= trans_data)
plt.title('Automatic vs Manual Cars Distribution')
plt.show()
```
